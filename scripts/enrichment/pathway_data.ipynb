```python
import requests
import time
```

```python


url = "http://rest.kegg.jp/link/pathway/hsa"
data = requests.get(url).text
gene_to_pathway_ids = {}
for line in data.strip().split("\n"):
    gene, pathway = line.split("\t")
    gene = gene.replace("hsa:", "")  # "7157"
    pathway = pathway.replace("path:", "")  # "hsa04110"
    if gene not in gene_to_pathway_ids:
        gene_to_pathway_ids[gene] = []
    gene_to_pathway_ids[gene].append(pathway)
print("DONE!")
```

```python
url = "http://rest.kegg.jp/link/pathway/hsa"
response = requests.get(url)
data = response.text
pathway_id_to_name = {}

for line in data.strip().split("\n"):
    pid, name = line.split("\t")
    pid = pid.replace("path:", "")  # hsa04110
    name = name.split(" - ")[0]  # remove species part
    pathway_id_to_name[pid] = name
print("DONE!")
```

```python
url = "http://rest.kegg.jp/list/hsa"
data = requests.get(url).text
gene_id_to_name = {}
for line in data.strip().split("\n"):
    line = line.split("\t")
    gid = line[0].replace("hsa:", "")
    gene_name = line[-1].split(";")[0]
    gene_id_to_name[gid] = gene_name
print("DONE!")
```

```python
gene_to_pathway_names = {}
for gene_id, pathways in gene_to_pathway_ids.items():
    gene_name = gene_id_to_name.get(gene_id, gene_id)
    pathway_names = []
    for pid in pathways:
        if pid in pathway_id_to_name:
            pathway_names.append(pathway_id_to_name[pid])

    gene_to_pathway_names[gene_name] = pathway_names
print("DONE!")
```

```python
import requests

def get_pathway_description(pathway_id):
    url = f"http://rest.kegg.jp/get/{pathway_id}"
    text = requests.get(url).text

    description = ""
    for line in text.split("\n"):
        if line.startswith("DESCRIPTION"):
            description = line.replace("DESCRIPTION", "").strip()
            break

    return description
```

```python
desc = get_pathway_description("hsa04110")
print(desc)
```

```python
pathway_id_to_description = {}
cache_dict = {}
for gene, pid in gene_to_pathway_ids.items():
    print(gene)
    for pathways in pid:
        print(pathways)
        pathway_id_to_description[gene] = []
        if pathways not in cache_dict:
            desc = get_pathway_description(pathways)
            cache_dict[pathways] = desc
            print(desc)
        else:
            print('fetching from cache')
            desc = cache_dict[pathways]
        pathway_id_to_description[gene].append(desc)
        time.sleep(0.5)

print('Done!')
```

```python
from transformers import AutoTokenizer, AutoModel
import torch

tokenizer = AutoTokenizer.from_pretrained("dmis-lab/biobert-v1.1")
model = AutoModel.from_pretrained("dmis-lab/biobert-v1.1")
model.eval()

def get_sentence_vector(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=512, padding=True)
    with torch.no_grad():
        outputs = model(**inputs)
    # Mean pool the token embeddings
    embedding = outputs.last_hidden_state.mean(dim=1).squeeze(0)
    return embedding

# Main loop
pathway_embeddings = {}

for pid, pathway_description in pathway_id_to_description.items():
    embeddings_list = []
    for description in pathway_description:
        embedding = get_sentence_vector(description)
        embeddings_list.append(embedding)

    embeddings = torch.stack(embeddings_list)
    final_vector = torch.mean(embeddings, dim=0)
    pathway_embeddings[pid] = final_vector
```

```python
ens_protein_ids_to_path_embeds = {}

def get_ensemble_ids(gene_id):
    response = requests.get(f"https://mygene.info/v3/gene/{gene_id}?fields=ensembl.protein")
    data = response.json()
    proteins = data.get("ensembl", {})

    # Normalize to always work with a list
    entries = proteins if isinstance(proteins, list) else [proteins]

    for entry in entries:
        if not isinstance(entry, dict):
            continue  # skip malformed entries
        prot = entry.get("protein")

        # protein can itself be a list (multiple isoforms for one transcript)
        if isinstance(prot, list):
            for p in prot:
                if p and p not in ens_protein_ids_to_path_embeds:
                    ens_protein_ids_to_path_embeds[p] = pathway_embeddings[gene_id]
        elif prot and prot not in ens_protein_ids_to_path_embeds:
            ens_protein_ids_to_path_embeds[prot] = pathway_embeddings[gene_id]
```

```python
for gene_id, embeddings in pathway_embeddings.items():
    print(gene_id)
    print(len(embeddings))
    print('protein_path_dict length ==>', len(ens_protein_ids_to_path_embeds))
    get_ensemble_ids(gene_id)
```
